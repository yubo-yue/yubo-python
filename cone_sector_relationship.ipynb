# 圆锥和扇形的关系

这个 notebook 用几何展开的方式说明：**圆锥的侧面展开后就是一个扇形**。

我们关注三个核心对应关系：

1. 扇形半径 = 圆锥母线长 `l`
2. 扇形弧长 = 圆锥底面周长 `2πr`
3. 扇形面积 = 圆锥侧面积 `πrl`


## 几何关系

设圆锥的：

- 底面半径为 `r`
- 高为 `h`
- 母线长为 `l`

则有：

$$
l = \sqrt{r^2 + h^2}
$$

圆锥侧面展开成扇形后：

- 扇形半径是 `l`
- 扇形弧长等于圆锥底面周长 `2\pi r`

因此扇形圆心角（弧度制）满足：

$$
\theta = \frac{2\pi r}{l}
$$

如果换成角度制，则：

$$
\theta = 360^\circ \cdot \frac{r}{l}
$$

扇形面积为：

$$
\frac{1}{2} l^2 \theta = \frac{1}{2} l^2 \cdot \frac{2\pi r}{l} = \pi r l
$$

这正好就是圆锥的侧面积。

```python
import numpy as np

# 选一个经典例子：3-4-5 直角三角形，对应圆锥 r=3, h=4, l=5
r = 3.0
h = 4.0
l = np.sqrt(r**2 + h**2)
theta = 2 * np.pi * r / l

arc_length = l * theta
base_circumference = 2 * np.pi * r
sector_area = 0.5 * l**2 * theta
lateral_area = np.pi * r * l

print(f"底面半径 r = {r:.2f}")
print(f"圆锥高 h = {h:.2f}")
print(f"母线长 l = {l:.2f}")
print(f"展开后扇形的半径 = {l:.2f}")
print(f"扇形圆心角 = {np.degrees(theta):.2f}°")
print(f"扇形弧长 = {arc_length:.2f}")
print(f"圆锥底面周长 = {base_circumference:.2f}")
print(f"扇形面积 = {sector_area:.2f}")
print(f"圆锥侧面积 = {lateral_area:.2f}")
```

## 图形对照

下面左图是圆锥，右图是它的侧面展开图。

观察重点：

- 圆锥顶点对应扇形圆心
- 圆锥母线对应扇形半径
- 圆锥底面圆周对应扇形弧线


```python
from IPython.display import HTML, display


def svg_text(x, y, text, color="#1f2937", size=16, anchor="middle"):
    return (
        f'<text x="{x}" y="{y}" fill="{color}" '
        f'font-size="{size}" text-anchor="{anchor}" '
        'font-family="Arial, sans-serif">'
        f"{text}</text>"
    )


def cone_svg() -> str:
    apex = (160, 35)
    center = (160, 220)
    rx, ry = 92, 24
    left = (center[0] - rx, center[1])
    right = (center[0] + rx, center[1])

    parts = [
        '<svg viewBox="0 0 320 280" width="320" height="280">',
        '<rect width="100%" height="100%" fill="#ffffff"/>',
        '<text x="160" y="24" text-anchor="middle" font-size="20" font-family="Arial, sans-serif">圆锥</text>',
        f'<line x1="{apex[0]}" y1="{apex[1]}" x2="{left[0]}" y2="{left[1]}" stroke="#2563eb" stroke-width="3"/>',
        f'<line x1="{apex[0]}" y1="{apex[1]}" x2="{right[0]}" y2="{right[1]}" stroke="#2563eb" stroke-width="3"/>',
        f'<ellipse cx="{center[0]}" cy="{center[1]}" rx="{rx}" ry="{ry}" fill="#bfdbfe" stroke="#1d4ed8" stroke-width="3"/>',
        f'<line x1="{apex[0]}" y1="{apex[1]}" x2="{center[0]}" y2="{center[1]}" stroke="#6b7280" stroke-width="2" stroke-dasharray="6 5"/>',
        f'<line x1="{center[0]}" y1="{center[1]}" x2="{right[0]}" y2="{right[1]}" stroke="#dc2626" stroke-width="2.5"/>',
        svg_text(apex[0], apex[1] - 8, '顶点', '#dc2626', 14),
        svg_text(center[0], (apex[1] + center[1]) / 2, '高 h', '#4b5563', 15, 'start'),
        svg_text((apex[0] + right[0]) / 2 + 14, (apex[1] + right[1]) / 2, '母线 l', '#1d4ed8', 15),
        svg_text((center[0] + right[0]) / 2, center[1] - 8, '底面半径 r', '#dc2626', 15),
        svg_text(center[0], center[1] + 48, '底面圆周 = 2πr', '#92400e', 15),
        '</svg>',
    ]
    return ''.join(parts)


def sector_svg(theta_radians: float) -> str:
    center = np.array([170.0, 210.0])
    radius = 145.0
    theta_deg = np.degrees(theta_radians)
    start_deg = -160.0
    end_deg = start_deg + theta_deg
    arc_angles = np.radians(np.linspace(start_deg, end_deg, 160))
    arc_points = [
        (center[0] + radius * np.cos(a), center[1] + radius * np.sin(a))
        for a in arc_angles
    ]
    points = [center] + arc_points + [center]
    points_str = ' '.join(f'{x:.1f},{y:.1f}' for x, y in points)

    start = arc_points[0]
    end = arc_points[-1]
    mid = arc_points[len(arc_points) // 2]

    parts = [
        '<svg viewBox="0 0 340 280" width="340" height="280">',
        '<rect width="100%" height="100%" fill="#ffffff"/>',
        '<text x="170" y="24" text-anchor="middle" font-size="20" font-family="Arial, sans-serif">展开后的扇形</text>',
        f'<polygon points="{points_str}" fill="#fed7aa" stroke="#ea580c" stroke-width="3"/>',
        f'<circle cx="{center[0]}" cy="{center[1]}" r="4" fill="#dc2626"/>',
        f'<line x1="{center[0]}" y1="{center[1]}" x2="{start[0]:.1f}" y2="{start[1]:.1f}" stroke="#ea580c" stroke-width="3"/>',
        f'<line x1="{center[0]}" y1="{center[1]}" x2="{end[0]:.1f}" y2="{end[1]:.1f}" stroke="#ea580c" stroke-width="3"/>',
        svg_text(center[0], center[1] + 22, '圆心', '#dc2626', 14),
        svg_text((center[0] + start[0]) / 2 - 10, (center[1] + start[1]) / 2, '半径 = l', '#9a3412', 15),
        svg_text(mid[0], mid[1] - 10, '弧长 = 2πr', '#92400e', 15),
        svg_text(center[0] + 30, center[1] - 18, f'θ = {theta_deg:.1f}°', '#7c2d12', 15, 'start'),
        '</svg>',
    ]
    return ''.join(parts)


html = f'''
<div style="display:flex; gap:24px; align-items:flex-start; flex-wrap:wrap;">
  <div>{cone_svg()}</div>
  <div>{sector_svg(theta)}</div>
</div>
'''

display(HTML(html))
```

```python
# 观察不同参数下扇形圆心角的变化
examples = [(2, 3), (3, 4), (4, 3), (5, 12)]

print("   r    h      l    θ(度)")
print("---------------------------")
for radius, height in examples:
    slant_height = np.sqrt(radius**2 + height**2)
    angle_deg = np.degrees(2 * np.pi * radius / slant_height)
    print(f"{radius:4.1f} {height:4.1f} {slant_height:6.2f} {angle_deg:8.2f}")
```

## 总结

圆锥和扇形的关系可以概括为：

- **圆锥侧面展开后得到一个扇形**
- **扇形半径等于圆锥母线长**
- **扇形弧长等于圆锥底面周长**
- **扇形面积等于圆锥侧面积**

如果你修改 `r` 和 `h`，就可以继续观察：圆锥越“瘦高”，母线长越大，展开后的扇形圆心角通常越小。
